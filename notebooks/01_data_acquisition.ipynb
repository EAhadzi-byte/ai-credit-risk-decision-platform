```python
# STANDARD LIBRARY
# __________________________________________

import os

import kagglehub

from pathlib import Path

import shutil

import pandas as pd
```

```python
# Download dataset from Kaggle
# __________________________________________

print("Downloading Lending Club dataset...")

download_path = Path(kagglehub.dataset_download("wordsforthewise/lending-club"))

print("Downloaded to:")
print(download_path)


# Find project root 

PROJECT_ROOT = Path.cwd()  

print("\nProject root:")

print(PROJECT_ROOT)


# Create data/raw directory

RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DATA_DIR.mkdir(parents=True, exist_ok=True)

print("\nSaving files to:")
print(RAW_DATA_DIR)


# Find CSV files

csv_files = [f for f in download_path.rglob("*.csv") if f.is_file()] 

print("\nCSV files found:")
for file in csv_files:
    print("-", file.name)

# Copy files to data/raw

for file in csv_files:
    destination = RAW_DATA_DIR / file.name

    if not destination.exists():
        shutil.copy(file, destination)
        print(f"Copied {file.name}")
    else:
        print(f"{file.name} already exists")

print("\n✓ Dataset successfully stored in data/raw")
```
